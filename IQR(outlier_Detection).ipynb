```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("placement.csv")
```

```python
df.head()
```

```python


plt.figure(figsize=(16,5))
plt.subplot(1,2,1)
sns.distplot(df['cgpa'])
plt.subplot(1,2,2)
sns.distplot(df['placement_exam_marks'])
plt.show()
```

```python
df["placement_exam_marks"].describe()
```

```python
sns.boxplot(data=df["placement_exam_marks"])
```

```python
#finding IQR
percentile_25th=df["placement_exam_marks"].quantile(0.25)
percentile_75th=df["placement_exam_marks"].quantile(0.75)
```

```python
IQR=percentile_75th-percentile_25th
print(IQR)
```

```python
upper_limit=percentile_75th+1.5*IQR
lower_limit=percentile_25th-1.5*IQR
```

```python
print("upper_limit",upper_limit)
print("lower_limit",lower_limit)
```

```python
#Outlier
df[df["placement_exam_marks"]>upper_limit]
```

```python
df[df["placement_exam_marks"]<lower_limit]
```

#Trimming


```python
new_df=df[df["placement_exam_marks"]<upper_limit]
new_df.shape
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Before trimming
plt.figure(figsize=(6, 4))

sns.boxplot(x=new_df["placement_exam_marks"])

plt.title("Placement Exam Marks")
plt.show()

plt.show()
```

#Cappping

```python
new_df_cap=df.copy()
new_df_cap["placement_exam_marks"]=np.where(
    new_df_cap["placement_exam_marks"]>upper_limit,
    upper_limit,
    np.where(
        new_df_cap["placement_exam_marks"]<lower_limit,
        lower_limit,
        new_df_cap["placement_exam_marks"]
    )
)
```

```python
new_df_cap["placement_exam_marks"].describe()
```

```python
sns.boxplot(x=new_df_cap["placement_exam_marks"])
plt.show()
```
