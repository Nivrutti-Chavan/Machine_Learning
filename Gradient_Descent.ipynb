#Own_Class

```python
class GD_Regressor:
  def __init__(self,learning_rate,ephochs):
    self.m=12500
    self.b=54654
    self.lr=learning_rate
    self.ephochs=ephochs
  def fit(self,X,y):
    for i in range(self.ephochs):
      slope_b = -2 * np.sum(y - self.m * X.ravel() - self.b)
      step_size_b=self.lr * slope_b
      self.b=self.b-step_size_b

      slope_m=-2*np.sum(y-self.m* X.ravel()-self.b * X.ravel())
      step_size_m=self.lr * slope_m
      self.m=self.m-step_size_m

    print("b____",self.b)
    print('m____',self.m)

```
