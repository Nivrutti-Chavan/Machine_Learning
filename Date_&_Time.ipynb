```python
import pandas as pd
import numpy as np
```

```python
date=pd.read_csv("orders.csv")
time=pd.read_csv("messages.csv")
```

```python
date.head()
```

```python
date["date"]=pd.to_datetime(date["date"])
date.info()
```

#1.Extract Year

```python
date["date_year"]=date["date"].dt.year
date.head(5)
```

#Extract Month

```python
date["date_month_no"]=date["date"].dt.month #for number
date["date_month_no"]=date["date"].dt.month_name() #for name


date.head()
```

#Extract Days

```python
date["Day"]=date["date"].dt.day_name()
date.head()
```

```python
time.head()
```

# Is weekend

```python
date["Is_weekend"]=np.where(date["Day"].isin(["Sunday","Saturday"]),True,False)
date.head()
```

# Time

```python
time["date"]=pd.to_datetime(time["date"])
time.info()
```

```python
time.head()
```

```python
time["Hours"]=time["date"].dt.hour
time["Minutes"]=time["date"].dt.minute
time["Seconds"]=time["date"].dt.second
time.head()
```

# Extract only time & Date

```python
time["Time"]=time["date"].dt.time
time["Date"]=time["date"].dt.date
time.head()
```

```python
import datetime
today=datetime.datetime.today()
time["Diff"]=(today- time['date']).dt.days
time.head()
```
